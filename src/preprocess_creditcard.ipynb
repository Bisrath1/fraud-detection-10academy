```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
import logging
import os

# Configure logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

try:
    # Create processed directory if it doesn't exist
    os.makedirs(r'C:\10x AIMastery\fraud-detection-10academy\data\processed', exist_ok=True)
    
    # Load creditcard.csv
    logging.info("Loading creditcard.csv...")
    creditcard_data = pd.read_csv(r'C:\10x AIMastery\fraud-detection-10academy\data\raw\creditcard.csv')
    
    # Verify columns
    logging.info("Columns in creditcard.csv: %s", creditcard_data.columns.tolist())

    # Preprocessing
    logging.info("Preprocessing creditcard.csv...")
    
    # Handle missing values and duplicates
    creditcard_data = creditcard_data.dropna(subset=['Class'])
    creditcard_data = creditcard_data.drop_duplicates()
    
    # Standardize numerical features
    num_cols = ['Time', 'Amount'] + [f'V{i}' for i in range(1, 29)]
    scaler = StandardScaler()
    creditcard_data[num_cols] = scaler.fit_transform(creditcard_data[num_cols])

    # Save processed data
    logging.info("Saving processed creditcard data...")
    creditcard_data.to_csv(r'C:\10x AIMastery\fraud-detection-10academy\data\processed\processed_creditcard_data.csv', index=False)
    logging.info("Preprocessing complete.")

except FileNotFoundError as e:
    logging.error(f"File not found: {e}")
    exit(1)
except Exception as e:
    logging.error(f"An error occurred during preprocessing: {e}")
    exit(1)
```
