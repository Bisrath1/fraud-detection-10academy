```python
import pandas as pd
import os

# Set working directory to project root
os.chdir(r'C:\10x AIMastery\fraud-detection-10academy')

# Load datasets
fraud_data = pd.read_csv('data/raw/Fraud_Data.csv')
ip_country = pd.read_csv('data/raw/IpAddress_to_Country.csv')
credit_data = pd.read_csv('data/raw/creditcard.csv')

# Check for missing values
print("🔍 Missing Values in Fraud_Data.csv:")
print(fraud_data.isnull().sum())

print("\n🔍 Missing Values in IpAddress_to_Country.csv:")
print(ip_country.isnull().sum())

print("\n🔍 Missing Values in creditcard.csv:")
print(credit_data.isnull().sum())
```

```python
# Check for duplicates
print("Fraud Data duplicates:", fraud_data.duplicated().sum())
print("IP Country duplicates:", ip_country.duplicated().sum())
print("Credit Card duplicates:", credit_data.duplicated().sum())

# Check data types
print("\nFraud Data types:\n", fraud_data.dtypes)
print("\nIP Country types:\n", ip_country.dtypes)
print("\nCredit Card data types:\n", credit_data.dtypes)
```

```python
# Drop duplicate rows from credit card data
credit_data.drop_duplicates(inplace=True)
print("✅ Duplicates removed. New shape:", credit_data.shape)
```

```python
# Convert time columns to datetime
fraud_data['signup_time'] = pd.to_datetime(fraud_data['signup_time'])
fraud_data['purchase_time'] = pd.to_datetime(fraud_data['purchase_time'])

# Convert IP addresses to integers
fraud_data['ip_address'] = fraud_data['ip_address'].astype('int64')
ip_country['lower_bound_ip_address'] = ip_country['lower_bound_ip_address'].astype('int64')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set styling
sns.set(style="whitegrid")

# 1. Class distribution
plt.figure(figsize=(6,4))
sns.countplot(data=fraud_data, x='class', palette='viridis')
plt.title('Fraud vs Non-Fraud Transactions (E-commerce)')
plt.xlabel('Class (0 = Legit, 1 = Fraud)')
plt.ylabel('Count')
plt.show()

# 2. Age distribution
plt.figure(figsize=(6,4))
sns.histplot(fraud_data['age'], bins=20, kde=True)
plt.title('Age Distribution')
plt.show()

# 3. Purchase value
plt.figure(figsize=(6,4))
sns.histplot(fraud_data['purchase_value'], bins=30, kde=True, color='orange')
plt.title('Purchase Value Distribution')
plt.show()

# 4. Categorical: Source
plt.figure(figsize=(6,4))
sns.countplot(data=fraud_data, x='source', order=fraud_data['source'].value_counts().index)
plt.title('Traffic Source Distribution')
plt.show()

# 5. Categorical: Browser
plt.figure(figsize=(6,4))
sns.countplot(data=fraud_data, x='browser', order=fraud_data['browser'].value_counts().index)
plt.title('Browser Distribution')
plt.xticks(rotation=45)
plt.show()
```

```python
# Class distribution
plt.figure(figsize=(6,4))
sns.countplot(data=credit_data, x='Class', palette='coolwarm')
plt.title('Fraud vs Non-Fraud Transactions (Credit Card)')
plt.xlabel('Class (0 = Legit, 1 = Fraud)')
plt.ylabel('Count')
plt.show()

# Amount distribution
plt.figure(figsize=(6,4))
sns.histplot(credit_data['Amount'], bins=40, kde=True, color='green')
plt.title('Transaction Amount Distribution')
plt.show()
```

```python
# 1. Purchase value vs class
plt.figure(figsize=(6,4))
sns.boxplot(data=fraud_data, x='class', y='purchase_value')
plt.title('Purchase Value by Class')
plt.show()

# 2. Age vs class
plt.figure(figsize=(6,4))
sns.boxplot(data=fraud_data, x='class', y='age')
plt.title('Age by Class')
plt.show()
```

```python
# Heatmap of V1-V28, Amount & Class
plt.figure(figsize=(14,10))
corr = credit_data.corr()
sns.heatmap(corr, cmap='coolwarm', annot=False, linewidths=0.5)
plt.title('Correlation Heatmap (Credit Card Data)')
plt.show()
```

```python
ip_country = ip_country.sort_values('lower_bound_ip_address').reset_index(drop=True)
```

```python
# Step 1: Sort both DataFrames
fraud_data_sorted = fraud_data.sort_values('ip_address')
ip_country_sorted = ip_country.sort_values('lower_bound_ip_address')

# Step 2: Merge using 'merge_asof' on ip_address and lower_bound_ip_address
merged = pd.merge_asof(
    fraud_data_sorted,
    ip_country_sorted,
    left_on='ip_address',
    right_on='lower_bound_ip_address',
    direction='backward'
)

# Step 3: Filter out rows where ip_address is beyond upper bound
merged = merged[merged['ip_address'] <= merged['upper_bound_ip_address']]
```

```python
print(merged[['ip_address', 'country']].head())
print("🌍 Countries mapped:", merged['country'].nunique())
```

```python
merged.to_csv('data/processed/fraud_data_with_country.csv', index=False)
```
